# Run paper code for fig.1 and fig.2

```python
import imp, sys, os
sys.path.append('../code/')
import nonlinear_modelv3
import pupil_dark_helpers as pdh
import numpy as np
import torch
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
```

```python
os.getcwd()
```

```python
# Set data path
data_path = os.getcwd().replace('analysis/notebooks', 'data/')
print(data_path)
save_path = os.getcwd().replace('analysis/notebooks', 'data/figure_data/')
print("save path: ", save_path)
```

## Random stim data

```python
# random stim data
db=[]
db.append({'mname':'FXa8','datexp':'2026_01_13','blk':'1'})
db.append({'mname':'FXa9','datexp':'2026_01_15','blk':'1'})
db.append({'mname':'FXa10','datexp':'2026_01_13','blk':'1'})
db.append({'mname':'LZ16','datexp':'2024_07_29','blk':'1'})
db.append({'mname':'TX139','datexp':'2024_08_10','blk':'1'})
db.append({'mname':'TX143','datexp':'2024_08_10','blk':'1'})
db.append({'mname':'TX144','datexp':'2024_08_10','blk':'1'})
dbs = np.array(db)
mouse_list = np.unique([entry['mname'] for entry in dbs])
mouse_list
```

```python
imp.reload(nonlinear_modelv3)
hyperparameters = {
            "learning_rate": 1e-3,
            "weight_decay": 1e-4,
            "batch_size": 1,
            "seed": 0,
            "segment_length": 2350, # 35 Hz; 61 seconds
            "n_latents": 256,
            "frac" : 0.2,
            "k_folds" : 5,
}
nonlinear_results_lightrandom, rastermap_out = nonlinear_modelv3.analyze_mouse_data(data_path, mouse_list, dbs, device, 'light_random', hyperparameters)
```

```python
np.save(f"{save_path}nonlinear_lightrandom_results_udcnv.npy", nonlinear_results_lightrandom)
np.save(f"{save_path}lightrandom_rastermap_out_udcnv.npy", rastermap_out)
```
